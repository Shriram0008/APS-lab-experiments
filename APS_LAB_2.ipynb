### **The Birthday Problem**

*In a group of people, what is the probability that at least two people share the same birthday.*

```python
def probability_no_shared_birthday(n):
    probability = 1.0
    for i in range(n):
        probability *= (365 - i) / 365
    return probability

people = [1, 2, 3, 4, 5, 10, 20, 23, 30]
print("People\tP(No Shared Birthday)\tP(Shared Birthday)")

for n in people:
    p_no_shared = probability_no_shared_birthday(n)
    p_shared = 1 - p_no_shared
    print(f"{n}\t{p_no_shared:.3f}\t\t\t{p_shared:.3f}")
```

```python
import matplotlib.pyplot as plt

people = range(1, 366)
probabilities = []
for n in people:
    probabilities.append(probability_no_shared_birthday(n))

plt.figure(figsize=(10, 6))
plt.plot(people, probabilities)

# p23 = probability_no_shared_birthday(23)
# plt.scatter(23, p23)
# plt.annotate(
#     f"23 people: {p23:.3f}",
#     (23, p23),
#     xytext=(40, 0.6),
#     arrowprops=dict(arrowstyle="->"))

plt.xlabel("Number of People")
plt.ylabel("Probability")
plt.title("Probability That No Two People Have the Same Birthday")
plt.grid(True)
plt.show()
```

```python
n = int(input("Enter the number of people: "))
print("\nNumber of people | P(No Shared Birthday) | P(Shared Birthday)")

for num_people in range(1, n + 1):
    probability_no_shared = 1.0
    for i in range(num_people):
        probability_no_shared *= (365 - i) / 365

    prob_shared = 1 - probability_no_shared

    print(f"{num_people:<17}| {probability_no_shared:<21.4f} | {prob_shared:<.4f}")
```
