#### **Problem 1**

A fair coin is tossed 100 times.

Calculate:
*   Number of heads  
*   Number of tails
*   Experimental probability of Head     
*   Experimental probability of Tail






```python
import random
tosses=[random.choice(["Head","Tail"]) for _ in range(100)]
heads=tosses.count("Head") # Count heads and tails
tails=tosses.count("Tail")
head_prob=heads/100 # Experimental probabilities
tail_prob=tails/100
print("Number of Heads:",heads)
print("Number of Tails:",tails)
print("Experimental Probability of Head:",head_prob)
print("Experimental Probability of Tail:",tail_prob)
```

### **Problem 2**

Check Effect of the number of trials.

Repeat the coin-toss experiment for: n=10, 100, 1,000, 10,000

```python
import random
trial=[10,100,1000,10000]
for n in trial:
  tosses=[random.choice(["Head","Tail"]) for _ in range(n)]
  heads=tosses.count("Head") # Count heads and tails
  tails=tosses.count("Tail")
  head_prob=heads/100 # Experimental probabilities
  tail_prob=tails/100
  print("Number of Trials:", n) # Output of each trial
  print("Heads:", heads)
  print("Tails:", tails)
  print("Experimental Probability of Head:",head_prob)
  print("Experimental Probability of Tail:",tail_prob)
  print()
```

### **Problem 4**

**Addition rule for disjoint events**

A die is rolled 10,000 times.

*Define:* A=getting 1, B= getting 6, check the events are joint or disjoint  

*Verify:* **P(A∪B)=P(A)+P(B)**

```python
import random
n = 10000
rolls=[random.randint(1,6)for _ in range(n)]
A=rolls.count(1)
B=rolls.count(6)
union=A+B

Prob_A=A/n
Prob_B=B/n
Prob_union=union/n

print("No.of.times A (getting 1):",A)
print("No.of.times B (getting 6):",B)
print("\nP(A):",Prob_A)
print("P(B):",Prob_B)
print("P(A U B):",Prob_union)
print("P(A) + P(B):",Prob_A+Prob_B)

if(Prob_union==(Prob_A+Prob_B)):
    print("\nAddition Rule Verified!")
else:
    print("\nAddition Rule Not Verified!")
```

### **Problem 5**

**Addition rule for overlapping events**

For one die roll,

*define:* A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}

*Verify:* **P(A∪B)=P(A)+P(B)−P(A∩B)**

```python
import random
n = 10000
rolls=[random.randint(1, 6)for _ in range(n)]
A=sum(1 for roll in rolls if roll in [2, 4, 6])
B=sum(1 for roll in rolls if roll in [4, 5, 6])
intersection= sum(1 for roll in rolls if roll in [4, 6])
union=sum(1 for roll in rolls if roll in [2,4,5,6])

Prob_A=A/n
Prob_B=B/n
Prob_intersection=intersection/n
Prob_union=union/n

print("No.of.times A (even):", A)
print("No.of.times B (>3):", B)
print("No.of.times A ∩ B:", intersection)
print("No.of.times A ∪ B:", union)

print("\nP(A):", Prob_A)
print("P(B):", Prob_B)
print("P(A ∩ B):", Prob_intersection)
print("P(A ∪ B):", Prob_union)

if(Prob_union==((Prob_A+Prob_B)-Prob_intersection)):
    print("\nAddition Rule Verified!")
else:
    print("\nAddition Rule Not Verified!")
```

### **Problem 6**

**Multiplication rule for independent events**

A coin is tossed and a die is rolled 10,000 times.

*Define:*
*   obtaining Head  
*   obtaining an even number

*Verify:* **P(A∩B)=P(A)P(B)**

```python
import random
n = 10000
tosses=[random.choice(["Head","Tail"]) for _ in range(n)]
rolls=[random.randint(1,6)for _ in range(n)]
A=tosses.count("Head")
B=sum(1 for roll in rolls if roll in [2, 4, 6])
intersection=sum(1 for i in range(n) if tosses[i]=="Head" and rolls[i] in [2, 4, 6])
Prob_A=A/n
Prob_B=B/n
Prob_intersection=intersection/n

print("Number of Heads:",A)
print("Number of Even numbers:",B)
print("Number of Head AND Even:",intersection)

print("\nP(A):",Prob_A)
print("P(B):",Prob_B)
print("P(A ∩ B):",Prob_intersection)

if(Prob_intersection==Prob_A*Prob_B):
    print("\nMultiplication Rule Verified!")
else:
    print("\nMultiplication Rule Not Verified!")
```
